```python
import os

def list_directory_contents(directory):
    if os.path.exists(directory):
        print(f"Inhalt des Ordners {directory}:")
        for item in os.listdir(directory):
            item_path = os.path.join(directory, item)
            if os.path.isdir(item_path):
                print(f"Verzeichnis: {item}")
            elif os.path.isfile(item_path):
                print(f"Datei: {item}")
            else:
                print(f"Anderes: {item}")
    else:
        print(f"Der Ordner {directory} existiert nicht.")

list_directory_contents("/home/lion")

list_directory_contents("/media/kobil/samsung41/lion")
```
