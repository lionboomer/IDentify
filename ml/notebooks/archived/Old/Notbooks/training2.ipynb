Schritt 1: Vorbereitung der Daten
Wir werden die sampleData-Spalte in numerische Werte umwandeln.

Schritt 2: Isolation Forest
Wir werden den Isolation Forest trainieren und die Ergebnisse anzeigen.


```python
import os
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
import base64

def connect_to_database():
    try:
        mongo_uri = os.getenv('MONGO_URI', 'mongodb://db:27017/fingerprintDB')
        client = pymongo.MongoClient(mongo_uri)
        db = client.get_default_database()
        print("Connected to the database.")
        return db
    except pymongo.errors.ConnectionError as err:
        print(f"Error: {err}")
        return None

# Funktion zur Umwandlung von Base64 in numerische Werte
def base64_to_numeric(base64_str):
    return np.frombuffer(base64.b64decode(base64_str), dtype=np.uint8)

# Funktion zur Berechnung statistischer Merkmale
def calculate_statistics(array):
    return {
        'mean': np.mean(array),
        'std': np.std(array),
        'min': np.min(array),
        'max': np.max(array),
        'sum': np.sum(array)
    }

# Funktion zur Erstellung eines Histogramms
def calculate_histogram(array, bins=10):
    hist, _ = np.histogram(array, bins=bins)
    return hist

# Daten vorbereiten
def prepare_data(df):
    df['numeric_sampleData'] = df['sampleData'].apply(base64_to_numeric)
    df['statistics'] = df['numeric_sampleData'].apply(calculate_statistics)
    df['histogram'] = df['numeric_sampleData'].apply(calculate_histogram)
    return df

# Isolation Forest Modell
def train_isolation_forest(df):
    # Überprüfen, ob die Liste nicht leer ist
    if df['numeric_sampleData'].empty:
        raise ValueError("Die numeric_sampleData-Spalte ist leer.")
    
    # Statistische Merkmale und Histogramme extrahieren
    stats_df = pd.DataFrame(df['statistics'].tolist())
    hist_df = pd.DataFrame(df['histogram'].tolist())
    
    # Kombinierte Merkmale
    X = pd.concat([stats_df, hist_df], axis=1)
    
    # Spaltennamen in Strings umwandeln
    X.columns = X.columns.astype(str)
    
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    
    model = IsolationForest(contamination=0.1)
    model.fit(X_scaled)
    
    return model, scaler

# Hauptfunktion zur Modellierung
def main():
    # Verbindung zur Datenbank herstellen und Daten laden
    db = connect_to_database()
    if db is None:
        print("Failed to connect to the database. Exiting...")
        exit()

    canvassamples_collection = db['canvassamples']
    canvassamples_cursor = canvassamples_collection.find()
    canvassamples_data = list(canvassamples_cursor)
    canvassamples_df = pd.DataFrame(canvassamples_data)

    fingerprints_collection = db['fingerprints']
    fingerprints_cursor = fingerprints_collection.find()
    fingerprints_data = list(fingerprints_cursor)
    fingerprints_df = pd.DataFrame(fingerprints_data)

    user_dataframes = {}
    for _, fingerprint in fingerprints_df.iterrows():
        user_id = fingerprint['_id']
        username = fingerprint['username']
        user_samples = canvassamples_df[canvassamples_df['fingerprintId'] == user_id]
        user_dataframes[username] = prepare_data(user_samples)

    # Isolation Forest Modelle für jeden Benutzer trainieren
    models = {}
    scalers = {}
    for username, df in user_dataframes.items():
        print(f"\nTraining Isolation Forest Modell für Benutzer: {username}")
        
        # Isolation Forest
        try:
            isolation_forest_model, scaler = train_isolation_forest(df)
            models[username] = isolation_forest_model
            scalers[username] = scaler
            print(f"Isolation Forest Modell für {username} trainiert.")
        except ValueError as e:
            print(f"Fehler beim Training des Modells für {username}: {e}")

if __name__ == "__main__":
    main()
```

```python
import numpy as np
import pandas as pd

def make_predictions(user_dataframes, models, scalers):
    predictions = {}
    for username, df in user_dataframes.items():
        print(f"\nVorhersagen für Benutzer: {username}")
        
        # Statistische Merkmale und Histogramme extrahieren
        stats_df = pd.DataFrame(df['statistics'].tolist())
        hist_df = pd.DataFrame(df['histogram'].tolist())
        
        # Kombinierte Merkmale
        X = pd.concat([stats_df, hist_df], axis=1)
        
        # Spaltennamen in Strings umwandeln
        X.columns = X.columns.astype(str)
        
        # Vorhersagen treffen
        X_scaled = scalers[username].transform(X)
        preds = models[username].predict(X_scaled)
        
        # Anomalien sind -1, normale Punkte sind 1
        preds = np.where(preds == -1, 0, 1)
        df.loc[:, 'anomaly'] = preds
        predictions[username] = df
        print(f"Anomalien für {username}:")
        print(df[df['anomaly'] == 0])
    
    return predictions

# Beispielaufruf (nach dem Training der Modelle)
# predictions = make_predictions(user_dataframes, models, scalers)
```

```python
from sklearn.metrics import confusion_matrix, f1_score, precision_score, recall_score

def evaluate_model(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    
    print("Confusion Matrix:")
    print(cm)
    print(f"F1 Score: {f1}")
    print(f"Precision: {precision}")
    print(f"Recall: {recall}")

def evaluate_predictions(predictions):
    for username, df in predictions.items():
        print(f"\nBewertung des Modells für Benutzer: {username}")
        
        y_true = np.ones(df.shape[0])  # Da es sich um Anomalieerkennung handelt, setzen wir alle Labels auf 1
        y_pred = df['anomaly'].values
        
        evaluate_model(y_true, y_pred)

# Beispielaufruf (nach dem Treffen der Vorhersagen)
# evaluate_predictions(predictions)
```

```python
if __name__ == "__main__":
    # Teil 1: Datenaufbereitung und Modelltraining
    main()
    
    # Teil 2: Vorhersagen treffen
    predictions = make_predictions(user_dataframes, models, scalers)
    
    # Teil 3: Modellbewertung
    evaluate_predictions(predictions)
```
